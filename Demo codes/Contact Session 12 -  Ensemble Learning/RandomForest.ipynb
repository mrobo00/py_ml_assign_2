# Random Forest


##  Task
Predict the maximum temperature of a day (`actual`) from the maximum temperatures of the previous two days (`temp_1`, `temp_2`), the historical average for that date (`average`), a friend's guess (`friend`) and calendar features.
This is a supervised, regression machine learning problem. It is Supervised as both the features (data for the city) and the targets (temperature) are available for training. It is Regression, because the target value is continuous.

**Random forest = bagging + random feature subsets.** Each tree is trained on a bootstrap sample of the training rows, and at every split it only considers a random subset of the features; the forest prediction is the *average* of the trees' predictions, which reduces the variance of a single deep tree.

Ref: https://towardsdatascience.com/random-forest-in-python-24d0893d51c0

## 1. Import the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

## 2. Prepare the dataset

```python
## Import the dataset
df = pd.read_csv('temps.csv')   # the CSV sits next to this notebook; change the path as necessary
print(df.head())
```

```python
# Set up the plotting layout
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize = (10,10))
fig.autofmt_xdate(rotation = 45)

# Actual max temperature measurement
ax1.plot(df['actual'])
ax1.set_xlabel('Day index'); ax1.set_ylabel('Temperature'); ax1.set_title('Max Temp')

# Temperature from 1 day ago
ax2.plot(df['temp_1'])
ax2.set_xlabel('Day index'); ax2.set_ylabel('Temperature'); ax2.set_title('Previous Max Temp')

# Temperature from 2 days ago
ax3.plot(df['temp_2'])
ax3.set_xlabel('Day index'); ax3.set_ylabel('Temperature'); ax3.set_title('Two Days Prior Max Temp')

# Friend Estimate
ax4.plot(df['friend'])
ax4.set_xlabel('Day index'); ax4.set_ylabel('Temperature'); ax4.set_title('Friend Estimate')

# This plot can be improved by grouping x-axis values in terms of month.
```

```python
# Perform One-Hot Encoding for categorical data.
# In this case, the column 'week'

features = pd.get_dummies(df)
features.head(7)
```

```python
# Convert to numpy array the training data and the target feature

X= features.drop('actual', axis = 1) # axis 1 refers to the columns
names = list(X.columns)
X = np.array(X)

y = np.array(features['actual'])
```

```python
# Split data into training and testing sets

(X_train, X_test, y_train, y_test) = train_test_split(X, y, test_size= 0.25, random_state=42)
```

```python
print('Training Features :', X_train.shape)
print('Training Labels :', y_train.shape)
print('Testing Features :', X_test.shape)
print('Testing Labels :', y_test.shape)
```

## 3. Build the Model

```python
# Instantiate model with 1000 decision trees
# max_depth=3 limits every tree to 8 leaves at most: each tree is weak and strongly regularised
# (try max_depth=None to grow full trees and compare the test error).
rf = RandomForestRegressor(n_estimators = 1000, random_state = 42, max_depth = 3)
```

```python
# Train the model on training data
rf.fit(X_train, y_train)
```

## 4. Make Predictions

```python
predictions = rf.predict(X_test)
```

## 5. Performance Measures

```python
# Calculate the absolute errors
errors = abs(predictions - y_test)

# Print out the mean absolute error (mae)
print('Mean Absolute Error:', round(np.mean(errors), 2), 'degrees.')
```

```python
# Calculate mean absolute percentage error (MAPE)
mape = 100 * (errors / y_test)
print('MAPE:', round(np.mean(mape), 2), '%')
# '100 - MAPE' is a handy rule of thumb, but it is NOT a classification-style accuracy
print('100 - MAPE:', round(100 - np.mean(mape), 2), '%')

# More standard regression measures
from sklearn.metrics import mean_squared_error, r2_score
print('RMSE:', round(np.sqrt(mean_squared_error(y_test, predictions)), 2), 'degrees')
print('R2  :', round(r2_score(y_test, predictions), 3))
```

## 6. Visualize random forest

```python
# Visualize one tree of the forest with sklearn's plot_tree (no graphviz / pydotplus needed)
from sklearn.tree import plot_tree
plt.figure(figsize=(22, 8))
plot_tree(rf.estimators_[3], feature_names=names, filled=True, rounded=True, precision=1)
plt.title('Decision Tree # 3')
plt.show()
```

```python
plt.figure(figsize=(22, 8))
plot_tree(rf.estimators_[995], feature_names=names, filled=True, rounded=True, precision=1)
plt.title('Decision Tree # 995')
plt.show()
```
